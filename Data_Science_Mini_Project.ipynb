```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10,6)

# 1. CREATE DATASET - You can also upload any Kaggle CSV here
data = {
    "Name": ["Shashikanth", "Sunny", "harshith", "Vijay", "Sweety",
             "Arjun", "Kiran", "Reena", "Ravi", "Pooja"],
    "Gender": ["F", "F", "F", "F", "F", "M", "M", "F", "M", "F"],
    "Hours_Studied": [8, 10, 5, 7, 6, 9, 4, 8, 6, 7],
    "Attendance": [95, 98, 75, 88, 80, 92, 70, 90, 78, 85],
    "Marks": [88, 92, 76, 85, 80, 90, 68, 87, 74, 82]
}
df = pd.DataFrame(data)

print("--- 1. DATASET OVERVIEW ---")
display(df.head())
print(df.info())
print("\nStatistical Summary:")
display(df.describe())

# 2. DATA ANALYSIS - Generate Insights
print("\n--- 2. KEY INSIGHTS ---")
print(f"Average Marks: {df['Marks'].mean():.2f}")
print(f"Highest Marks: {df['Marks'].max()} by {df.loc[df['Marks'].idxmax(), 'Name']}")
print(f"Lowest Marks: {df['Marks'].min()} by {df.loc[df['Marks'].idxmin(), 'Name']}")
print(f"Correlation between Hours Studied and Marks: {df['Hours_Studied'].corr(df['Marks']):.2f}")

# 3. VISUALIZATIONS

# Visualization 1: Bar Chart - Marks by Student
plt.figure()
sns.barplot(x="Name", y="Marks", data=df, palette="viridis")
plt.title("Student Marks Comparison")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
# Insight: akhila scored highest, kiran scored lowest

# Visualization 2: Scatter Plot - Hours Studied vs Marks
plt.figure()
sns.scatterplot(x="Hours_Studied", y="Marks", hue="Gender", data=df, s=100)
plt.title("Hours Studied vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.grid(True)
plt.tight_layout()
plt.show()
# Insight: Positive correlation. More hours = Higher marks

# Visualization 3: Histogram - Distribution of Attendance
plt.figure()
sns.histplot(df["Attendance"], bins=5, kde=True, color="orange")
plt.title("Distribution of Student Attendance")
plt.xlabel("Attendance %")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
# Insight: Most students have 80-90% attendance

# Visualization 4: Box Plot - Marks by Gender
plt.figure()
sns.boxplot(x="Gender", y="Marks", data=df, palette="pastel")
plt.title("Marks Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Marks")
plt.tight_layout()
plt.show()
# Insight: Female students have slightly higher median marks

# 4. SUMMARY OF FINDINGS
print("\n--- 4. SUMMARY OF FINDINGS ---")
print("1. There is a strong positive correlation between Hours Studied and Marks.")
print("2. Attendance also impacts performance. Students with >85% attendance scored above 80.")
print("3. The top performer is akhila with 92 marks.")
print("4. Gender-wise, female students in this sample performed slightly better on average.")
```
